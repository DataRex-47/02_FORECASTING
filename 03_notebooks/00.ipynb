```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

historico = pd.DataFrame({'evento': ['Sin_evento', 'Christmas', 'Halloween', 'Sin_evento']})
hoy = pd.DataFrame({'evento': ['Sin_evento', 'Sin_evento']})

# Encoder que aprendió del histórico
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ohe.fit(historico)
print(ohe.transform(hoy))         # salen 3 columnas

# Encoder que aprende solo con lo de hoy
ohe_nuevo = OneHotEncoder(sparse_output=False)
print(ohe_nuevo.fit_transform(hoy))   # sale 1 columna
```
